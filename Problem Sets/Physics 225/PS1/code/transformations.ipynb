```python
import numpy as np 
import matplotlib.pyplot as plt

plt.rcParams.update({
    'text.usetex': True,
    'font.family': 'lmodern',
})
```

```python
t = np.linspace(-20, 20, 500)
x = np.linspace(-20, 20, 500)

v = 0.5

# Worldlines
r = np.array([0*x, t])
u = np.array([v*t, t])

# Lorentz transformation
k = 1
A = 1 / np.sqrt( 1 - k * v**2)

L = A * np.array([
    [1, -v],
    [- k* v,  1]
])

rp = L @ r
up = L @ u

# Spacing
xspacing = 1
tspacing = 1

fig, ax = plt.subplots(1, 2, figsize=(6, 3), dpi=300)
# r and u
ax[0].plot(*r, 'r-',  lw=1.5, zorder = 2)
ax[0].plot(*u, 'b-',  lw=1.5, zorder = 2)

# r' and u'
ax[1].plot(*rp, 'r-', lw=1.5, zorder = 2)
ax[1].plot(*up, 'b-', lw=1.5, zorder = 2)

# light lines

l1 = np.array([ c * t, t])
l2 = np.array([-c * t, t])

ax[0].plot(*l1,  color = "#FFFF00", ls = '-', lw=1, zorder = 1)
ax[0].plot(*l2,  color = "#FFFF00", ls = '-', lw=1, zorder = 1)
ax[1].plot(*(L@l1),  color = "#FFFF00", ls = '-', lw=1, zorder = 1)
ax[1].plot(*(L@l2),  color = "#FFFF00", ls = '-', lw=1, zorder = 1)

for x0 in np.arange(x.min(), x.max(), xspacing):
    line = np.array([
        x0 * np.ones_like(t),
        t
    ])

    ax[0].plot(
        *line,
        'k-', lw=0.3,
        zorder = 0
    )
    line_p = L @ line

    ax[1].plot(
        *line_p,
        'k-', lw=0.3,
        zorder = 0
    )

    if x0 == 0:
        ax[0].plot(
            *line,
            'k-', lw=0.75,
            zorder = 0
        )
        ax[1].plot(
            *line_p,
            'k-', lw=0.75,
            zorder = 0
        )

for t0 in np.arange(t.min(), t.max(), tspacing):
    line = np.array([
        x,
        t0 * np.ones_like(x)
    ])

    ax[0].plot(
        *line,
        'k-', lw=0.3,
        zorder = 0
    )
    line_p = L @ line

    ax[1].plot(
        *line_p,
        'k-', lw=0.3,
        zorder = 0
    )

    if t0 == 0:
        ax[0].plot(
            *line,
            'k-', lw=0.75,
            zorder = 0
        )
        ax[1].plot(
            *line_p,
            'k-', lw=0.75,
            zorder = 0
        )

# ==================================================
# Formatting
# ==================================================

for axs in ax:
    axs.set_aspect('equal')
    axs.set_xlim(-5, 5)
    axs.set_ylim(-5, 5)
    axs.set_xlabel('$x$')
    axs.set_ylabel('$t$')

plt.tight_layout()
plt.savefig(r'..\images\k1.pdf')
plt.show()
```
